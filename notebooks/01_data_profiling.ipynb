```python
import os
import pandas as pd
from pathlib import Path

# Garante o caminho absoluto correto independente de onde o notebook é executado
# Sobe um nível a partir de notebooks/ para chegar na raiz do projeto
BASE_DIR = Path.cwd().parent if Path.cwd().name == "notebooks" else Path.cwd()
SAMPLES_DIR = BASE_DIR / "data" / "samples"

# Cria a pasta caso ela não exista
SAMPLES_DIR.mkdir(parents=True, exist_ok=True)

PATH_BASICS = SAMPLES_DIR / "sample_basics.tsv"
PATH_RATINGS = SAMPLES_DIR / "sample_ratings.tsv"
PATH_EPISODES = SAMPLES_DIR / "sample_episodes.tsv"

# URLs oficiais do IMDb
URLS = {
    PATH_BASICS: "https://datasets.imdbws.com/title.basics.tsv.gz",
    PATH_RATINGS: "https://datasets.imdbws.com/title.ratings.tsv.gz",
    PATH_EPISODES: "https://datasets.imdbws.com/title.episode.tsv.gz"
}

# Se os arquivos locais não existirem, baixa 1.000 linhas de cada direto da web
for local_path, url in URLS.items():
    if not local_path.exists():
        print(f"Baixando amostra para: {local_path.name}...")
        df_sample = pd.read_csv(url, sep="\t", nrows=1000, na_values="\\N")
        df_sample.to_csv(local_path, sep="\t", index=False)
        print(f"-> Salvo em {local_path}")
    else:
        print(f"Amostra já existe localmente: {local_path.name}")

# Carrega os dataframes locais
df_basics = pd.read_csv(PATH_BASICS, sep="\t", na_values="\\N")
df_ratings = pd.read_csv(PATH_RATINGS, sep="\t", na_values="\\N")
df_episodes = pd.read_csv(PATH_EPISODES, sep="\t", na_values="\\N")

print("\nTudo pronto! As 3 amostras estão carregadas na memória.")
```
